# Super-Cortex // Spectral Graph Theory & Topological Manifold Analytics
Fetches live graph components and quantitative scorecard from Super-Cortex,
analyzing the Graph Laplacian spectrum, Fiedler algebraic connectivity, and Louvain communities.


```python
import httpx
import networkx as nx
import polars as pl

CORTEX_HOST = 'http://127.0.0.1:4174'
SCOPE = 'all'

# 1. Fetch full topological summary
summary = httpx.get(f'{CORTEX_HOST}/api/analytics/summary', params={'scope': SCOPE}).json()
print("=== SUPER-CORTEX GRAPH TOPOLOGY SCORECARD ===")
print(f"Nodes: {summary['node_count']} | Edges: {summary['edge_count']}")
print(f"Modularity Q: {summary['modularity']:.4f}")
print(f"Algebraic Connectivity λ₂: {summary['algebraic_connectivity']:.6f}")
print(f"Average Clustering Coefficient: {summary['avg_clustering_coefficient']:.4f}")
```

```python
# 2. Ingest into NetworkX DiGraph
nx_data = httpx.get(f'{CORTEX_HOST}/api/export/networkx', params={'format': 'json', 'scope': SCOPE}).json()
G = nx.node_link_graph(nx_data, edges='links')
print(f"Constructed NetworkX graph: {G}")
```
