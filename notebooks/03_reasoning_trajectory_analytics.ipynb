# Super-Cortex // AI Reasoning Trajectory & Fine-Tuning Dataset Hub
Inspects Super-Cortex multi-agent reasoning traces, thinking paths, and DPO margin distributions.


```python
import httpx
import json
import polars as pl

CORTEX_HOST = 'http://127.0.0.1:4174'
SCOPE = 'all'

# Fetch SFT and DPO datasets from Super-Cortex
sft_resp = httpx.get(f'{CORTEX_HOST}/api/export/trajectories', params={'format': 'sft', 'scope': SCOPE})
dpo_resp = httpx.get(f'{CORTEX_HOST}/api/export/trajectories', params={'format': 'dpo', 'scope': SCOPE})

sft_data = sft_resp.json()['dataset']
dpo_data = dpo_resp.json()['dataset']

print(f"Loaded {len(sft_data)} SFT samples and {len(dpo_data)} DPO pairs.")
```

```python
# Convert DPO pairs to Polars DataFrame for reward analysis
df = pl.DataFrame([
    {
        'id': p['id'],
        'chosen_reward': p['chosen']['reward'],
        'rejected_reward': p['rejected']['reward'],
        'margin': p['margin'],
        'has_thinking': bool(p['chosen'].get('thinking')),
    }
    for p in dpo_data
])
print(df.describe())
```
